```python
df_clean = df[df['Reach'] > 0].copy()
```

```python
df_clean.shape
```

```python
df_clean['Total_Interactions'] = (
    df_clean['Likes']
    + df_clean['Comments']
    + df_clean['Shares']
    + df_clean['Saves']
)
```

```python
df_clean[['Content_ID', 'Likes', 'Comments', 'Shares', 'Saves', 'Total_Interactions']].head()
```

```python
df_clean['Engagement_Rate'] = (
    df_clean['Total_Interactions'] / df_clean['Reach']
) * 100
```

```python
df_clean[['Content_ID', 'Reach', 'Total_Interactions', 'Engagement_Rate']].head()
```

```python
df_clean['IPR'] = (
    df_clean['Total_Interactions'] / df_clean['Reach']
)
```

```python
####Menghitung IPI

##IPI = Interaction Per Impression

df_clean['IPI'] = (
    df_clean['Total_Interactions'] / df_clean['Impressions']
)
```

```python
df_clean['IPI_Percent'] = df_clean['IPI'] * 100
```

```python
##Menghitung CTR

##CTR = Click Through Rate.
df_clean['CTR_Calc'] = (
    df_clean['Clicks'] / df_clean['Impressions']
) * 100
```

```python
###Membuat Tabel Final Metrik
metrics = df_clean[
    [
        'Content_ID',
        'Platform',
        'Content_Type',
        'Reach',
        'Impressions',
        'Likes',
        'Comments',
        'Shares',
        'Saves',
        'Clicks',
        'Total_Interactions',
        'Engagement_Rate',
        'IPR',
        'IPI',
        'CTR_Calc'
    ]
]
```

```python
metrics.head()
```

```python
###Mengurutkan Berdasarkan Engagement Rate
ranking_er = metrics.sort_values(
    by='Engagement_Rate',
    ascending=False
)
```

```python
ranking_er[
    [
        'Content_ID',
        'Platform',
        'Content_Type',
        'Reach',
        'Total_Interactions',
        'Engagement_Rate'
    ]
].head(10)
```

```python
##Menentukan Konten Terbaik
ranking_er.iloc[0]
```

```python
ranking_er.iloc[0]['Content_ID']
```

```python
##Menentukan 10 Konten Terendah
ranking_er[
    [
        'Content_ID',
        'Platform',
        'Content_Type',
        'Reach',
        'Total_Interactions',
        'Engagement_Rate'
    ]
].tail(10)
```

```python
###Analisis Platform
platform_analysis = df_clean.groupby('Platform').agg(
    Total_Content=('Content_ID', 'count'),
    Avg_Reach=('Reach', 'mean'),
    Avg_Impressions=('Impressions', 'mean'),
    Avg_Interactions=('Total_Interactions', 'mean'),
    Avg_ER=('Engagement_Rate', 'mean'),
    Avg_IPI=('IPI', 'mean'),
    Avg_CTR=('CTR_Calc', 'mean')
).reset_index()
```

```python
platform_analysis.sort_values(
    'Avg_ER',
    ascending=False
)
```

```python
###Analisis Content Type
content_type_analysis = df_clean.groupby('Content_Type').agg(
    Total_Content=('Content_ID', 'count'),
    Avg_Reach=('Reach', 'mean'),
    Avg_Impressions=('Impressions', 'mean'),
    Avg_Interactions=('Total_Interactions', 'mean'),
    Avg_ER=('Engagement_Rate', 'mean'),
    Avg_IPI=('IPI', 'mean'),
    Avg_CTR=('CTR_Calc', 'mean')
).reset_index()
```

```python
content_type_analysis.sort_values(
    'Avg_ER',
    ascending=False
)
```

```python
content_type_analysis
```

```python
promotion_analysis = df_clean.groupby('Is_Promoted').agg(
    Total_Content=('Content_ID', 'count'),
    Avg_Reach=('Reach', 'mean'),
    Avg_Interactions=('Total_Interactions', 'mean'),
    Avg_ER=('Engagement_Rate', 'mean'),
    Avg_CTR=('CTR_Calc', 'mean')
).reset_index()
```

```python
###Membandingkan Promoted vs Organic
promotion_analysis
```

```python
###Melihat Korelasi
numeric_cols = [
    'Reach',
    'Impressions',
    'Likes',
    'Comments',
    'Shares',
    'Saves',
    'Clicks',
    'Total_Interactions',
    'Engagement_Rate',
    'IPI',
    'CTR_Calc'
]

df_clean[numeric_cols].corr()
```

```python
###Visualisasi Ranking Engagement
top10 = ranking_er.head(10).sort_values(
    'Engagement_Rate'
)

plt.figure(figsize=(10,6))

plt.barh(
    top10['Content_ID'],
    top10['Engagement_Rate']
)

plt.xlabel('Engagement Rate (%)')
plt.ylabel('Content ID')
plt.title('Top 10 Content berdasarkan Engagement Rate')

plt.show()
```

```python
###Visualisasi Platform
platform_plot = platform_analysis.sort_values(
    'Avg_ER',
    ascending=True
)

plt.figure(figsize=(10,6))

plt.barh(
    platform_plot['Platform'],
    platform_plot['Avg_ER']
)

plt.xlabel('Average Engagement Rate (%)')
plt.ylabel('Platform')
plt.title('Average Engagement Rate per Platform')

plt.show()
```

```python
###Visualisasi Content Type
type_plot = content_type_analysis.sort_values(
    'Avg_ER',
    ascending=True
)

plt.figure(figsize=(10,6))

plt.barh(
    type_plot['Content_Type'],
    type_plot['Avg_ER']
)

plt.xlabel('Average Engagement Rate (%)')
plt.ylabel('Content Type')
plt.title('Average Engagement Rate per Content Type')

plt.show()
```
